```python
import os                         # Helps interact with file system.
from PIL import Image             # Used to open and display image file easily.
import matplotlib.pyplot as plt   # For plotting images and text.
import random                     # To pick random images to check variety and correctness of data loading.
```

```python
# Storing the paths to Images folder and Caption folder

IMAGE_DIR = "../data/raw/Flickr8k_Dataset/Images"
CAPTION_FILE = "../data/raw/Flickr8k_Dataset/captions.txt"


print("Number of images: ", len(os.listdir(IMAGE_DIR)))
print("Captions file exists: ", os.path.exists(CAPTION_FILE))
```

```python
captions_dict = {}

with open(CAPTION_FILE, 'r') as f:
    for line in f:
        line = line.strip()
        if not line:
            continue
        image_id, caption = line.split('\t')
        image_id = image_id.split(',')[0]
        if image_id not in captions_dict:
            captions_dict[image_id]=[]
        captions_dict[image_id].append(caption)
```
